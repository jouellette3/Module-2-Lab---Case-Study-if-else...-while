```python
"""
Jonathan Ouellette
testStudentGPA.ipynb
This app checks to see if a student's grade point average is high enough to be on 
the Dean's list and/or Honor ROLL
"""

while True:

    student_lastname = input("Enter a student's last name, Enter 'ZZZ' to exit the program: ")

    if student_lastname == 'ZZZ' or 'zzz':
        break

    student_firstname = input("Enter a student's first name: ")

    student_grade = float(input("Enter the student's GPA: "))

    if student_grade >= 3.5:
        print("This student has made the Dean's List!")
    if student_grade >= 3.25:
        print("This student has made the Honor Roll!")
        continue
    else:
        print("This student has neither made the Dean's List or Honor Roll.")
```
